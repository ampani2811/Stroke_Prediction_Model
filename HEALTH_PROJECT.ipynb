```python
import pandas as pd
```

```python
#Load you data from source
data=pd.read_csv('healthcare-dataset-stroke-data.csv')
```

```python
#view first twenty(20) rows of the data
data.head(20)
```

```python
#total numbers of columns
data.columns
```

```python
#bmi column has missing values
print(data['bmi'])
```

```python
#Data cleaning by applying the mean to all missing values in bmi column
data['bmi']= data['bmi'].fillna(data['bmi'].mean())
```

```python
#view first five(5) rows after the cleaaning the data
data.head(5)
```

```python
data.head(10)
```

```python
# To covert categorical data for proper analysis
data=pd.get_dummies(data, drop_first = True)
```

```python
data.head(5)
```

```python
#Training and testing the model
x=data.drop('stroke', axis=1)
y=data['stroke']
from sklearn.model_selection import train_test_split
x_train,x_test, y_train, y_test = train_test_split(x,y, test_size=0.2, random_state=42)
```

```python
#model development
from sklearn.linear_model import LogisticRegression
model=LogisticRegression(max_iter =10000)
model.fit(x_train, y_train)
```

```python
pred_y=model.predict(x_test)
```

```python
print("Result of the Predicted Model :", pred_y)
```

```python
print(y_test)
```

```python
print(y_train)
```

```python
#To check for accuracy of the result
from sklearn.metrics import accuracy_score
accuracy=accuracy_score(y_test, pred_y)
```

```python
print("Accuracy:", accuracy)
```
